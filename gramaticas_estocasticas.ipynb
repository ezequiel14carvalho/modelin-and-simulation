```python
import numpy as np

grammar = {
    "S": [(1.0, "Eu", "V")],
    "V": [(0.6, "gosto de", "O"), (0.4, "como", "O")],
    "O": [(0.5, "maçã", "F"), (0.3, "banana", "F"), (0.2, "laranja", "F")],
    "F": [(1.0, None, None)]
}

def gerar_frase(grammar, estado_inicial = "S"):
    estado_atual = estado_inicial
    frase = []

    while estado_atual is not None:
        trans = grammar[estado_atual]
        prob = [p for p, _, _, in trans]
        escolha = np.random.choice(len(trans), p=prob)
        estado_atual = trans[escolha][2]

        if trans[escolha][1] is not None:
            frase.append(trans[escolha][1])
        
    return " ".join(frase)

for _ in range(14):
    print(gerar_frase(grammar))
```
