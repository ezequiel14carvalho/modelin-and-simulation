Queda Livre

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.lines import lineStyles


def analitica_v(g, v0, T):
    vel = np.arange(0, T+1, 1)
    for t in vel:
        vel[t] = v0 - g*t

    return vel


def analitica_y(g, v0, y0, t):
    pos = np.arange(0, t+1, 1)
    for t in pos:
        pos[t] = 1/2*(-g*t**2) + v0*t + y0

    return pos

def euler_queda_livre(g, v0, y0, dt, t):
    t_values = np.arange(0, t + dt, dt)
    vel = np.zeros_like(t_values)
    pos = np.zeros_like(t_values)

    pos[0], vel[0] = y0, v0

    for i in range(1, len(pos)):
        vel[i] = vel[i-1] - g*dt
        pos[i] = pos[i-1] + vel[i-1]*dt

    return t_values, vel, pos

# Parâmetros
g = 9.81
v0 = 0
y0 = 100
dt = 0.5
T = 5
tempos = np.arange(0, T+1, 1)

# Quedra livre analítico
velAnalitica = analitica_v(g, v0, T)
posAnalitica = analitica_y(g, v0, y0, T)

# Queda livre Euler
t_values, velEuler, posEuler = euler_queda_livre(g, v0, y0, dt, T)

# Plotagem
fig, ax = plt.subplots(1, 2)
ax.plot(tempos, posAnalitica, linestyle='--', label='Solução Analítica', color='red')
ax.plot(t_values, posEuler, linestyle='--', label='Solução Euler', color='blue')
ax.legend()
ax.set_title("Comparação da distância")
plt.xlabel("Tempo (s)")
plt.ylabel("Distância (m)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
