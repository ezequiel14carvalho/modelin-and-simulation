# Main

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

# fase 1
# aceleração constante positiva

# fase 2
# velocidade constante

# fase 3
# desacelerção constante até parar

a_0 = 2
v_0 = 10

t_1 = v_0/a_0
t_2 = 2*t_1
t_3 = t_2 + v_0/a_0

def x_t(t):
  ''' (float) -> (float)
      Recebe um instante de tempo e retorna a posição do drone correspondente
      a esse instante.
  '''
  # fase 1
  if t < t_1:
    x1 = (1/2)*(a_0)*(t)**2
    return x1

  # fase 2
  elif t < t_2:
    x1 = (1/2)*(a_0)*(t_1)**2
    x2 = (v_0)*(t - t_1) +  x1
    return x2

  # fase 3
  else:
    x1 = (1/2)*(a_0)*(t_1)**2
    x2 = x1 + v_0*(t_2 - t_1)
    x3 = -(1/2)*a_0*(t - t_2)**2 + v_0*(t-t_2) + x2
    return x3

def v_t(t):
  ''' (float) -> (float)
      Recebe um instante de tempo e retorna a velocidade do drone correspondente
      a esse instante.
  '''
  # fase 1
  if t < t_1:
    v_1 = a_0*t
    return v_1

  # fase 2
  elif t < t_2:
    return v_0

  # fase 3
  else:
    return -a_0*(t - t_2) + v_0


def a_t(t):
  ''' (float) -> (float)
      Recebe um instante de tempo e retorna a aceleração do drone correspondente
      a esse instante.
  '''
  # fase 1
  if t < t_1:
    return a_0

  # fase 2
  elif t < t_2:
    return 0

  # fase 3
  else:
    return -a_0


def criar_graficos_pos(ini,fim,n):
  ''' (float,float,int) -> (None)
      Recebe um o inicio, o fim e a quantidade de instantes de tempo e plota o gráfico
      da posição pelo tempo do drone nesse intervalo, criando instantes de tempo
      uniformemente distribuídos.
  '''
  tempo = np.linspace(ini,fim,)

  res_x = []

  for i in tempo:
    res_x.append(x_t(i))


  print(res_x)

  fig, ax = plt.subplots(figsize = (10,5), layout = "constrained")
  ax.plot(tempo,res_x, color="red")
  ax.set_xlabel("Tempo(s)")
  ax.set_ylabel("Espaço (m)")
  ax.set_title("Posição X Tempo")

  plt.show()

def criar_graficos_vel(ini,fim,n):
  ''' (float,float,int) -> (None)
      Recebe um o inicio, o fim e a quantidade de instantes de tempo e plota o gráfico
      da velocidade pelo tempo do drone nesse intervalo, criando instantes de tempo
      uniformemente distribuídos.
  '''
  tempo = np.linspace(ini,fim,)

  res_x = []

  for i in tempo:
    res_x.append(v_t(i))


  print(res_x)

  fig, ax = plt.subplots(figsize = (10,5), layout = "constrained")
  ax.plot(tempo,res_x, color="red")
  ax.set_xlabel("Tempo(s)")
  ax.set_ylabel("Velocidade (m/s)")
  ax.set_title("velocidade X Tempo")

  plt.show()


def criar_graficos_acel(t_1,t_2,t_3,n):
  ''' (float,float,float,int) -> (None)
      Recebe instantes de tempo que correspondem ao incio e fim dos intervalos de teste, bem como o número de instantes para o teste
      e plota o gráfico da posição pelo tempo do drone nesse intervalo, criando instantes de tempo
      uniformemente distribuídos.
  '''
  # A função da aceleração de plotagem é diferente das de velociade e posição pois
  # quis quebrar o gráfico em barras não conexas.

  int_1 = np.linspace(0,t_1-0.01)
  int_2 = np.linspace(t_1,t_2-0.01)
  int_3 = np.linspace(t_2,t_3)

  res_1  = []
  res_2  = []
  res_3  = []

  fig, ax = plt.subplots(figsize = (10,5), layout = "constrained")

  for t in int_1:
    res_1.append(a_t(t))

  ax.plot(int_1,res_1,label=f'[0:{t_1:.1f}]',color = "red")

  for t in int_2:
    res_2.append(a_t(t))

  ax.plot(int_2,res_2,label=f'[{t_1:.1f}:{t_2:.1f}]',color = "red")

  for t in int_3:
    res_3.append(a_t(t))

  ax.plot(int_3,res_3,label=f'[{t_2:.1f}:{t_2:.1f}]',color = "red")
  ax.set_xlabel("Tempo(s)")
  ax.set_ylabel("Aceleração (m/s^2)")
  ax.set_title("Aceleração X Tempo")
  plt.show()
```
