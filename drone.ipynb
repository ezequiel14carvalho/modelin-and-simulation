```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros
a0 = 2.0
v0 = 100
t1 = v0/a0
t2 = 2*t1
t3 = t1 + t2
x1 = 0.5 * a0 * t1**2
x2 = v0 * (t2 - t1) + 0.5 * a0 * t1**2

def vel(T):
    v_t = np.zeros_like(T)

    for i, t in enumerate(T):
        if t < t1:
            v_t[i] = a0*t
        elif t < t2:
            v_t[i] = v0
        else:
            v_t[i] = v0 - a0*(t - t2)

    return v_t

def pos(T):
    x_t = np.zeros_like(T)

    for i, t in enumerate(T):
        if t < t1:
            x_t[i] = 1/2*a0*t**2
        elif t < t2:
            x_t[i] = v0*(t - t1) + x1
        else:
            x_t[i] = v0*(t - t2) + x2 - 1/2*a0*(t - t2)**2

    return x_t

def ace(T):
    a_t = np.zeros_like(T)

    for i, t in enumerate(T):
        if t < t1:
            a_t[i] = a0
        elif t < t2:
            a_t[i] = 0
        else:
            a_t[i] = -a0

    return a_t

# Vetor de tempo
t = np.linspace(0, t3, 10000)

pos_values = pos(t)
vel_values = vel(t)
ace_values = ace(t)

# Plotagem
fig, ax = plt.subplots(3, 1, figsize=(10, 8))

ax[0].plot(t, pos_values)
ax[0].set_title('Posição vs Tempo')
ax[0].set_xlabel('Tempo (s)')
ax[0].set_ylabel('Posição (m)')
ax[0].grid(True)
ax[0].set_xticks([0, t1, t2, t3])

ax[1].plot(t, vel_values)
ax[1].set_title('Velocidade vs Tempo')
ax[1].set_xlabel('Tempo (s)')
ax[1].set_ylabel('Velocidade (m/s)')
ax[1].grid(True)
ax[1].set_xticks([0, t1, t2, t3])

ax[2].plot(t, ace_values)
ax[2].set_title('Aceleração vs Tempo')
ax[2].set_xlabel('Tempo (s)')
ax[2].set_ylabel('Aceleração (m/s²)')
ax[2].grid(True)
ax[2].set_xticks([0, t1, t2, t3])

plt.tight_layout()
plt.show()
```
