```python
import numpy as np
from matplotlib import pyplot as plt

def mapa_logistico(r, x0, N):
    resultados = [x0]
    for i in range(N):
        x0 = r * x0 * (1 - x0)
        resultados.append(x0)
    return resultados

# Teste 1: Trajetoria para ponto fixo e ponto de bifurcação
r = [2.7, 3.1]
x0 = [0.1, 0.2]
x_fix = [(r[0] - 1) / r[0], (r[1] - 1) / r[1]]
x_max_distance = [np.abs(x_fix[0] - x0[0]), np.abs(x_fix[1] - x0[1])]
N = [30, 30]
resultados = [mapa_logistico(r[0], x0[0], N[0]), mapa_logistico(r[1], x0[1], N[1])]

fig, ax = plt.subplots(1, 2, figsize=(14, 5))

ax[0].plot(resultados[0], marker='o', linestyle='--', label='Iterações do Mapa Logístico')
ax[0].set_title(f'Mapa Logístico (Estabilidade Fixa): r={r[0]}, x0={resultados[0][0]:.2f}')
ax[0].set_xlabel('Iteração')
ax[0].set_ylabel('Valor de x')
ax[0].set_ylim(x_fix[0] - x_max_distance[0], x_fix[0] + x_max_distance[0])
ax[0].axhline(y=x_fix[0], color='r', linestyle='--', label=f'Ponto Fixo: {x_fix[0]:.2f}')
ax[0].grid()
ax[0].legend()

ax[1].plot(resultados[1], marker='o', linestyle='--', label='Iterações do Mapa Logístico')
ax[1].set_title(f'Mapa Logístico (Estabilidade Periódica): r={r[1]}, x0={resultados[1][0]:.2f}')
ax[1].set_xlabel('Iteração')
ax[1].set_ylabel('Valor de x')
ax[1].set_ylim(x_fix[1] - x_max_distance[1], x_fix[1] + x_max_distance[1])
ax[1].grid()
ax[1].legend()

plt.tight_layout()
plt.show()

#Teste 2: Diagrama de bifurcação
r_values = np.linspace(2.8, 4.0, 1000)
x0 = 0.1
N = 500
trasiente = 400
plt.figure(figsize=(14, 7))

for r in r_values:
    resultados = mapa_logistico(r, x0, N)
    estavel = resultados[trasiente:]
    r_plot = [r] * len(estavel)
    plt.plot(r_plot, estavel, marker=',', linestyle='', color='black', markersize=0.5)
plt.title('Diagrama de Bifurcação do Mapa Logístico')
plt.xlabel('Taxa de Crescimento (r)')
plt.ylabel('Valor de x')
plt.grid()
plt.show()
        
```
