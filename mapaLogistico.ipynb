```python
import numpy as np
from matplotlib import pyplot as plt

def mapa_logistico(r, x):
    return r * x * (1 - x)

# Parâmetros
r = 2.8
x0 = 0.1
x_fix = (r - 1) / r
x_max_distance = np.abs(x_fix - x0)
resultados = [x0]
N = 30

for i in range(N):
    x0 = mapa_logistico(r, x0)
    resultados.append(x0)

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(resultados, marker='o', linestyle='--', label='Iterações do Mapa Logístico')
ax.set_title(f'Mapa Logístico: r={r}, x0={resultados[0]:.2f}')
ax.set_xlabel('Iteração')
ax.set_ylabel('Valor de x')
ax.set_ylim(x_fix - x_max_distance, x_fix + x_max_distance)
ax.axhline(y=x_fix, color='r', linestyle='--', label=f'Ponto Fixo: {x_fix:.2f}')
ax.grid()
ax.legend()

plt.tight_layout()
plt.show()
```
