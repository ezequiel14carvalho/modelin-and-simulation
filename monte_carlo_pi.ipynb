Estimação do π usando Monte Carlo

```python
import numpy as np
import matplotlib.pyplot as plt

N = [10, 100, 1000, 10000, 100000, 1000000]
estimates_pi = []
erros = []
true_pi = np.pi

for n in N:
    x = np.random.uniform(0, 1, n)
    y = np.random.uniform(0, 1, n)

    inside = x**2 + y**2 <= 1
    pi_estimate = 4 * np.mean(inside)
    estimates_pi.append(pi_estimate)
    erros.append(abs(pi_estimate - true_pi))


fig, ax = plt.subplots(1, 2, figsize=(12, 5))
# Estimativa de π
ax[0].plot(N, estimates_pi, marker='o')
ax[0].axhline(true_pi, color='r', linestyle='--', label='True π')
ax[0].set_xscale('log')
ax[0].set_xlabel('Number of Points (log scale)')
ax[0].set_ylabel('Estimated π')
ax[0].set_title('Estimation of π using Monte Carlo')    
ax[0].legend()
# Erro da estimativa
ax[1].plot(N, erros, marker='o')
ax[1].set_xscale('log')
ax[1].set_xlabel('Number of Points (log scale)')
ax[1].set_ylabel('Absolute Error')
ax[1].set_title('Error of π Estimation')

plt.tight_layout()
plt.show()
```
